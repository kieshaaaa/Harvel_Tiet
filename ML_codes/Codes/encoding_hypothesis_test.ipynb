```python
import pandas as pd
from scipy.stats import ranksums
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = '/content/data_sheet of_BPE_wordpiece_pos.csv'
data = pd.read_csv(file_path)

# Perform Wilcoxon rank-sum test
def perform_test(group1, group2, metric):
    stat, p_value = ranksums(group1, group2)
    print(f"{metric}: Statistic={stat:.4f}, p-value={p_value:.4e}")

# WER Hypothesis Testing
print("\nWER Comparisons:")
perform_test(data['BPE_WER'], data['WordPiece_WER'], 'BPE vs WordPiece WER')
perform_test(data['BPE_WER'], data['Positional_WER'], 'BPE vs Positional WER')
perform_test(data['WordPiece_WER'], data['Positional_WER'], 'WordPiece vs Positional WER')

# BLEU Hypothesis Testing
print("\nBLEU Comparisons:")
perform_test(data['BPE_BLEU'], data['WordPiece_BLEU'], 'BPE vs WordPiece BLEU')
perform_test(data['BPE_BLEU'], data['Positional_BLEU'], 'BPE vs Positional BLEU')
perform_test(data['WordPiece_BLEU'], data['Positional_BLEU'], 'WordPiece vs Positional BLEU')

# Prepare data for visualization
wer_data = data[['BPE_WER', 'WordPiece_WER', 'Positional_WER']].melt(var_name='Technique', value_name='WER')
bleu_data = data[['BPE_BLEU', 'WordPiece_BLEU', 'Positional_BLEU']].melt(var_name='Technique', value_name='BLEU')

# Plot WER Scores
plt.figure(figsize=(12, 6))
sns.boxplot(x='Technique', y='WER', data=wer_data, palette="Set2")
sns.swarmplot(x='Technique', y='WER', data=wer_data, color=".25", alpha=0.7)
plt.title("WER Score Distribution Across Encoding Techniques", fontsize=16)
plt.ylabel("WER Score", fontsize=12)
plt.xlabel("Encoding Technique", fontsize=12)
plt.show()

# Plot BLEU Scores
plt.figure(figsize=(12, 6))
sns.boxplot(x='Technique', y='BLEU', data=bleu_data, palette="Set3")
sns.swarmplot(x='Technique', y='BLEU', data=bleu_data, color=".25", alpha=0.7)
plt.title("BLEU Score Distribution Across Encoding Techniques", fontsize=16)
plt.ylabel("BLEU Score", fontsize=12)
plt.xlabel("Encoding Technique", fontsize=12)
plt.show()

# Optional: Mean bar plots for summary
plt.figure(figsize=(12, 6))
wer_means = wer_data.groupby('Technique')['WER'].mean().reset_index()
sns.barplot(x='Technique', y='WER', data=wer_means, palette="Set2")
plt.title("Mean WER Scores Across Encoding Techniques", fontsize=16)
plt.ylabel("Mean WER Score", fontsize=12)
plt.xlabel("Encoding Technique", fontsize=12)
plt.show()

plt.figure(figsize=(12, 6))
bleu_means = bleu_data.groupby('Technique')['BLEU'].mean().reset_index()
sns.barplot(x='Technique', y='BLEU', data=bleu_means, palette="Set3")
plt.title("Mean BLEU Scores Across Encoding Techniques", fontsize=16)
plt.ylabel("Mean BLEU Score", fontsize=12)
plt.xlabel("Encoding Technique", fontsize=12)
plt.show()
```

```python
import pandas as pd
from scipy.stats import ranksums
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_path = '/content/data_sheet of_BPE_wordpiece_pos.csv'  # Update this with the correct file path
data = pd.read_csv(file_path)

# Perform Wilcoxon rank-sum test
def perform_test(group1, group2, metric):
    stat, p_value = ranksums(group1, group2)
    print(f"{metric}: Statistic={stat:.4f}, p-value={p_value:.4e}")

# WER Hypothesis Testing
print("\nWER Comparisons:")
perform_test(data['BPE_WER'], data['WordPiece_WER'], 'BPE vs WordPiece WER')
perform_test(data['BPE_WER'], data['Positional_WER'], 'BPE vs Positional WER')
perform_test(data['WordPiece_WER'], data['Positional_WER'], 'WordPiece vs Positional WER')

# BLEU Hypothesis Testing
print("\nBLEU Comparisons:")
perform_test(data['BPE_BLEU'], data['WordPiece_BLEU'], 'BPE vs WordPiece BLEU')
perform_test(data['BPE_BLEU'], data['Positional_BLEU'], 'BPE vs Positional BLEU')
perform_test(data['WordPiece_BLEU'], data['Positional_BLEU'], 'WordPiece vs Positional BLEU')

# Prepare data for visualization
wer_data = data[['samples', 'BPE_WER', 'WordPiece_WER', 'Positional_WER']]
bleu_data = data[['samples', 'BPE_BLEU', 'WordPiece_BLEU', 'Positional_BLEU']]

# Line Plot for WER Scores
plt.figure(figsize=(14, 7))
plt.plot(wer_data['samples'], wer_data['BPE_WER'], marker='o', label='BPE WER', linestyle='-', color='blue')
plt.plot(wer_data['samples'], wer_data['WordPiece_WER'], marker='s', label='WordPiece WER', linestyle='--', color='green')
plt.plot(wer_data['samples'], wer_data['Positional_WER'], marker='d', label='Positional WER', linestyle='-.', color='red')
plt.title("WER Scores Across Samples", fontsize=16)
plt.xlabel("samples", fontsize=12)
plt.ylabel("WER Score", fontsize=12)
plt.legend(fontsize=12)
plt.grid(True)
plt.show()

# Line Plot for BLEU Scores
plt.figure(figsize=(14, 7))
plt.plot(bleu_data['samples'], bleu_data['BPE_BLEU'], marker='o', label='BPE BLEU', linestyle='-', color='blue')
plt.plot(bleu_data['samples'], bleu_data['WordPiece_BLEU'], marker='s', label='WordPiece BLEU', linestyle='--', color='green')
plt.plot(bleu_data['samples'], bleu_data['Positional_BLEU'], marker='d', label='Positional BLEU', linestyle='-.', color='red')
plt.title("BLEU Scores Across Samples", fontsize=16)
plt.xlabel("samples", fontsize=12)
plt.ylabel("BLEU Score", fontsize=12)
plt.legend(fontsize=12)
plt.grid(True)
plt.show()
```
